```python
#Setup of api and of folder where data is being stored 
import os 
import fastf1

os.makedirs('cache', exist_ok=True)
fastf1.Cache.enable_cache('cache')
```

```python
#Just a test to see if everything works 
my_session = fastf1.get_session(2025, 'Australian Grand Prix', 'Q')
my_session.load ()
```

```python
my_laps = my_session.laps
my_laps.head()
```

```python
my_session.results[['Abbreviation', 'FullName']] #Just to make sure the correct abbreviations are being used later on
```

```python
my_laps_AntvsRus = my_laps.pick_drivers(['ANT','RUS'])
my_laps_AntvsRus 
```

```python
my_antonelli_fastestlap = my_laps_AntvsRus.pick_drivers('ANT').pick_fastest()
my_russell_fastestlap = my_laps_AntvsRus.pick_drivers('RUS').pick_fastest()

print(my_antonelli_fastestlap[['Driver','LapTime']])
print(my_russell_fastestlap[['Driver','LapTime']])
```

```python
my_gap = my_antonelli_fastestlap['LapTime'] - my_russell_fastestlap['LapTime']
my_gap
```
